## AA vs. SIE

```python
from Anderson import Anderson, get_norm_vs_time
from SIE import SIE
import matplotlib.pyplot as plt
from Colors import Colors, nice_grid, nice_legend
import numpy as np

global_dpi = 300


t = 0.5
m1: Anderson = Anderson().get_from_pkl(file='cluster_results/aa_m1_i5.pkl')
m2: Anderson = Anderson().get_from_pkl(file='cluster_results/aa_m2_i5.pkl')
m3: Anderson = Anderson().get_from_pkl(file='cluster_results/aa_m3_i5.pkl')
m4: Anderson = Anderson().get_from_pkl(file='cluster_results/aa_m4_i5.pkl')
sie: SIE = SIE().get_from_pkl(file='cluster_results/sie_i14_t1.pkl')

plt.figure(figsize=(5,3), dpi=600)
plt.plot(m1.x[t][-1], lw=0.5, label=f'$aa_{1}$')
plt.plot(m2.x[t][-1], lw=0.5, label=f'$aa_{2}$')
plt.plot(m3.x[t][-1], lw=0.5, label=f'$aa_{3}$')
plt.plot(m4.x[t][-1], lw=0.5, label=f'$aa_{4}$')
plt.plot(sie.x[t][-1], 'ks--', markerfacecolor='none', markersize=4, lw=0.5, mew=0.5)
nice_grid()
nice_legend()

plt.figure(figsize=(5,3), dpi=600)
n1 = get_norm_vs_time(res=m1, ref=sie, order=2)
n2 = get_norm_vs_time(res=m2, ref=sie, order=2)
n3 = get_norm_vs_time(res=m3, ref=sie, order=2)
n4 = get_norm_vs_time(res=m4, ref=sie, order=2)
plt.plot(sie.times, n1, label=f'$aa_{1}$')
plt.plot(sie.times, n2, label=f'$aa_{2}$')
plt.plot(sie.times, n3, label=f'$aa_{3}$')
plt.plot(sie.times, n4, label=f'$aa_{4}$')
nice_grid()
nice_legend()
print(f"avg. norm1 = {np.mean(n1)}")
print(f"avg. norm2 = {np.mean(n2)}")
print(f"avg. norm3 = {np.mean(n3)}")
print(f"avg. norm4 = {np.mean(n4)}")
```

```python
t = 0.5

plt.figure(figsize=(5,3), dpi=600)
lab = 0
for this in sie.x[t]:
  lab += 1
  plt.plot(this, lw=0.5,label=f"$x_{lab}$")
nice_grid()
nice_legend()

plt.figure(figsize=(5,3), dpi=600)
lab = 0
for this in sie.fx[t]:
  lab += 1
  plt.plot(this, lw=0.5,label=f"$f(x_{lab})$")
nice_grid()
nice_legend()

```
