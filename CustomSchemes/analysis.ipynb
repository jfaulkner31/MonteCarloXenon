```python
import SIE
import matplotlib.pyplot as plt
```

```python
the_file = SIE.SIE().get_from_pkl("../CustomSchemesAnalysis/SIE_Scheme/FLUX.pkl")
plt.figure(figsize=(5,3))
for time in the_file.fx.keys():
  plt.plot(the_file.fx[time][-1])

plt.figure(figsize=(5,3))
t1,n1=the_file.get_Nmat_by_time(nuc='Xe135', mat_name='11')
t2,n2=the_file.get_Nmat_by_time(nuc='Xe135', mat_name='13')
t3,n3=the_file.get_Nmat_by_time(nuc='Xe135', mat_name='15')
plt.plot(t1[3:],n1[3:])
plt.plot(t2[3:],n2[3:])
plt.plot(t3[3:],n3[3:])
```
